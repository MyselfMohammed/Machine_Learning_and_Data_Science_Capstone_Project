```python
import pandas as pd
import requests
import time

# API endpoint
url = "https://api.coingecko.com/api/v3/coins/markets"

# Parameters that stay constant
params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 250,  # Max allowed per page
    "sparkline": False,
}

all_data = []

# Loop through 40 pages (40 * 250 = 10,000 records)
for page in range(1, 41):
    print(f"Fetching page {page}...")
    params["page"] = page
    response = requests.get(url, params=params)

    # Check for API limit or error
    if response.status_code != 200:
        print(f"Error on page {page}: {response.status_code}")
        break

    data = response.json()

    if not data:  # Break if no data returned
        break

    all_data.extend(data)
    time.sleep(1)  # Respectful delay between requests

# Convert to DataFrame
df = pd.DataFrame(all_data)

# Save to Excel
df.to_excel("crypto_10000_records.xlsx", index=False)
print("Excel file saved as: crypto_10000_records.xlsx")
```
